```python
ewbins = np.arange(round(table['ew'].min(), 1), round(table['ew'].max(), 1)+0.4, 0.2)
spectra_ids = np.unique(table['spectrum_id'])
fracs = []
velocities = []
snrs = []
ews = []

for id in spectra_ids:
    w = table['spectrum_id'] == id
    rows = table[w]

    nreal = np.sum(w)
    frac = np.sum(rows['p'] >= 0.95) / nreal
    fracs.append(frac)
    velocities.append(np.mean(rows['v_synth']))
    snrs.append(np.mean(rows['SNR']))
    ews.append(np.mean(rows['ew']))

    # if frac == 1 and np.mean(rows['v_synth']) == 20:
    #     print(id)

# thresholds = []
# for i in range(len(ewbins)-1):
#     ewmin = ewbins[i]
#     ewmax = ewbins[i+1]
#     binthresh = []
#     for sn in np.unique(snrs):
#         wsn = sn == np.array(snrs)
#         wew = (np.array(ews) >= ewmin) & (np.array(ews) < ewmax)
#         w = wsn & wew
#         v = np.array(velocities)[w]
#         f = np.array(fracs)[w]

#         cut = f == 1
#         binthresh.append(np.mean(v[cut]))
#     thresholds.append(binthresh)

thresholds = {}
for sn in np.unique(snrs):
    wsn = sn == np.array(snrs)
    ew = ews[wsn]
    ewbins = np.arange(round(ew.min(), 1), round(ew.max(), 1) + 0.4, 0.4)

    thresholds[f"{sn}"] = {}
    for i in range(len(ewbins) -1):
        wew = (ews >= ewbins[i]) & (ews < ewbins[i+1])

        w = wsn & wew
        v = np.array(velocities)[w]
        f = np.array(fracs)[w]

        cut = f == 1
        thresh = np.mean(v[cut])
        
        # thresholds[f"{sn}"]["ewrange"] = (ewbins[i], ewbins[i+1])
        # thresholds[f"{sn}"]["vthresh"] = thresh

        thresholds[f"{sn}"][f"{ewbins[i]} <= EW <{ewbins[i+1]}"] = thresh


for snstr, subdict in thresholds.items():
    print(f"S/N = {snstr}")
    for ewkey, vtresh in subdict.items():
        print(f"    {ewkey} : v = {vtresh}")
    

# for i in range(len(thresholds)):
#     print(f"{ewbins[i]} <= EW < {ewbins[i+1]}")
#     binthresh = thresholds[i]
#     for sn, v in zip(np.unique(snrs), binthresh):
#         print(f"    S/N {sn}: v = {v}")
```
